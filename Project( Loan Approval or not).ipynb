## 1. Buisiness Understanding

#### Problem Statement

A bank wants to Predict whether a loan application should be approved based on the applicant's

personal and financial information. This helps reduce manual effort and speeds up the loan approval process.


#### Buisiness Objectives

* Reduce Manual verification
* Minimize the Risk of Approval high-Risk applicants
* Improve Customer experience

#### Machine Learning Problem

* Type: Supervised Learning
* Category: Binary Classification
* Target Variable: loan_status

## 2. Dataset Understanding

These are the main Features/Columns available in the dataset :

1. No_of_dependents – Number of dependents the applicant has
2. Education – Graduate / Not Graduate
3. Self_employed – Whether the applicant is self-employed
4. Income_annum – Annual income of the applicant
5. Loan_amount – Loan amount requested
6. Loan_term – Loan repayment time (years)
7. Cibil_score – Credit score indicating creditworthiness
8. Residential_assets_value – Value of residential assets
9. Commercial_assets_value – Value of commercial assets
10. Luxury_assets_value – Value of luxury assets
11. Bank_asset_value – Bank balance / financial assets
12. Loan_status (Target variable) – Approved (1) or Rejected (0)

Target:
* Approved
* Rejected

## 10. Hyperparameter Tuning

Hyperparameters tuning is the process of finding best parameter values for a ML Model

to improve its performance.

* Improve Model accuracy
* Reduce overfitting
* Find the best model automatically

#### Method1: GridsearchCV

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier
```

```python
# Parameters Grid

param_grid={
    'n_estimators':[50,100,200],
    'max_depth':[3,10,15],
    'min_samples_split':[2,5,10]
}
```

```python
# Apply Grid

rf=RandomForestClassifier(random_state=42)


grid=GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy'
    
)

grid.fit(x_train,y_train)
```

```python
# Best parameters

print('Best Parameters:',grid.best_params_)
```

```python
print('Best Score:',grid.best_score_)
```

```python
# Best Model

best_rf=grid.best_estimator_
y_pred=best_rf.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print('Test Accuracy:',accuracy_score(y_test,y_pred))
```

#### Method 2: RandomizedSearchCV

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
# parameters

param_dist={
    'n_estimators':[50,100,150,200,250,300],
    'max_depth':[5,10,15,20,None],
    'min_samples_split':[2,5,10],
    'min_samples_leaf':[1,2,4]
}
```

```python
# Apply Model

random_search=RandomizedSearchCV(
    estimator=rf,
    param_distributions=param_dist,
    n_iter=10,
    cv=5,
    scoring='accuracy',
    random_state=42
)
```

```python
# Best parameters

random_search.fit(x_train,y_train)
```

```python
# Best Model
best_model=random_search.best_estimator_
y_pred=best_model.predict(x_test)
```

```python
print('Best Parameters:',random_search.best_params_)
```

```python
print('Accuracy:',accuracy_score(y_test,y_pred))
```
